## tuple, set, dictionary 실습 숙제

byte, bytes, bytearray 공부

all(), any()

```python
# 변경불가
tuple1 = (1, 2, 3, 3)
print(tuple1)

# 변경 방법
fix_tuple1 = list(tuple1)
fix_tuple1.append(4)
tuple1 = tuple(fix_tuple1)
print(tuple1)

# 중복제거
set1 = {1, 2, 3, 3, 4}
print(set1)

# 교,합집합
set2 = {3, 4, 5}
print(set1 & set2)
print(set1 | set2)

# 초기 딕셔너리 세팅
dict1 = {'name': 'kai', 'age': 20}
print("age:", dict1['age'])

# 키,값 추가
dict1['age'] = 21
dict1['grade'] = 'A'
print(dict1)
```

```python
a = b"Hello"
print("바이트:", a)
print("첫 바이트", a[0]) 

a = bytearray(b"Hello")
print("원본:", a)

# H를A에 해당하는 바이트로 변경
a[0] = 65 
print("수정", a)
```

```python
list_a = [True, True, 1, "Hello"]
list_b = [True, False, 0, ""]
list_c = [False, 0, ""]

# all은 and처럼 모든 요소가 참일때 참
print("all(list_a):", all(list_a)) 
print("all(list_b):", all(list_b)) 

# any는 or처럼 하나라도 참이면 참
print("any(list_b):", any(list_b)) 
print("any(list_c):", any(list_c)) 
```
