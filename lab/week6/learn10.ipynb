# 작업폴더 python 파일 실행하는 위치

```python
with open("text.txt", "w") as f:
    f.write("Hello, World!")
```

```python
with open("text.txt", "r", encoding="utf-8") as f:
    #for line in f:
    #    print(line.strip())

    lines = f.readline().strip() 
    while lines!= "":
        print(lines)
        lines = f.readline().strip()

    eline = f.readline().strip()
    print(eline)
    print(f.tell())

    f.seek(0, 0)
    eline = f.readline().strip()
    print(eline)
    print(f.tell())

    
```

# 전체 읽기
while 한줄저장객체 != "": # 비었을때까지 가져오는코드

# 첫 줄 제거
next(한줄저장객체, None(반환값))
