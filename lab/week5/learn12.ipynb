# 12장 상속

- 상속에 대한 이론과 실제 학습

```python
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMake(self, maker):
        self.maker = maker

    def getMaker(self):
        return self.maker

    def getDesc(self):
        return self.__str__()

    def __str__(self):
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"


class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterysize = bat_size

    def getBatterySize(self):
        return self.batterysize

    def setBatterySize(self, bat_size):
        self.batterysize = bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery: {self.batterysize}kWh"

    def getDesc(self):
        return super().getDesc()


class Truck(Car):
    def __init__(self, maker, model, color, price, payload):
        super().__init__(maker, model, color, price)
        self.payload = payload

    def __str__(self):
        return f"{super().__str__()}, Payload: {self.payload}"


car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.getDesc())
print(car1.model)

ecar1 = ElectricCar("Tesla", "Model S", "Red", 8000, 81)
print(ecar1)
ecar1.setBatterySize(92)
print(ecar1.getDesc())

truck1 = Truck("Ford", "F-150", "Blue", 4000, 1000)
print(truck1)
print(truck1.getMaker())

print(type(truck1))
print(type(ecar1))
print(type(car1))

print(isinstance(truck1, Truck))
print(isinstance(ecar1, ElectricCar))
print(isinstance(car1, Car))

print(isinstance(truck1, Car))
print(isinstance(ecar1, Car))



```

# 속성변수의 가시성: public, protected(_), praivate(__)

```python
class Parent(object):
    def __init__(self):
        self.__private_money = 1000
        self._protected_money = 500
        self.public_money = 100

    def __str__(self):
        return f"Parent has total {self.__private_money+self._protected_money+self.public_money} money"

class Child(Parent):
    def __init__(self):
        super().__init__()
        
    def __str__(self):
        return f"Child: {self._protected_money + self.public_money} money"   

def test():
    p = Parent()
    print(p)
    c = Child()
    print(c)
    print(p.public_money)
    print(p._protected_money)
#    print(p.__private_money)

    print(f"Child : {c.public_money}")
    print(f"Child : {c._protected_money}")
#    print(f"Child : {c.__private_money}")

test()
```

# 다중 상속: 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self, v1 = 10):
        self.attri1 = v1

    def display1(self):
        print(f"Base1: {self.attri1}")
    
    def display(self):
        self.display1()

class Base2:
    def __init__(self, v2 = 20):
        self.attri2 = v2

    def display2(self):
        print(f"Base2: {self.attri2}")

    def display(self):
        self.display2()

class C(Base1, Base2):
    def __init__(self, v1, v2, v3):
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.attri3 = v3

    def print(self):
        print(f"{self.attri1}, {self.attri2}, {self.attri3}")
        self.display1()
        self.display2()
        Base1.display(self)
        Base2.display(self)
        
def test():
    c = C(v1=1, v2=2, v3=3)
    c.print()

    c2 = C(4, 5, 6)
    c2.print()

if __name__ == "__main__":
    test()
```

Overriding: 부모의 

```python
import math

class Shape:
    def __init__(self, name="도형"):
        self.name = name

    def draw(self):
        print(f"Shape: {self.name}을 그립니다.")

    def get_area(self):
        print(f"Shape: {self.name}의 면적을 계산합니다.")

class Circle(Shape):
    def __init__(self, radius=1, name="원"):
        super().__init__(name)
        self.radius = radius
        
    def draw(self):
        print(f"Circle: {self.name}을 그립니다.")

    def get_area(self) -> float:
        print(f"Circle: {self.name}의 면적을 계산합니다.")
        return math.pi * self.radius ** 2
    
class Rectangle(Shape):
    def __init__(self, width=1, height=1, name="사각형"):
        super().__init__(name)
        self.width = width
        self.height = height

    def draw(self):
        print(f"Rectangle: {self.name}을 그립니다.")

    def get_area(self) -> float:
        print(f"Rectangle: {self.name}의 면적을 계산합니다.")
        return self.width * self.height

def test():
    c = Circle(10)
    r1 = Rectangle(width=5, height=10)
    c.draw()
    print(c.get_area())

    r1.draw()
    print(r1.get_area())
    print("--------------------")

    circle = [Circle(1), Circle(2), Circle(3)]
    for c in circle:
       # c.draw()
        print(c.get_area())
        print("--------------------")

    print("---------------------")
    for r in [Rectangle(1, 1), Rectangle(2, 2), Rectangle(3, 3)]:
        r.draw()
        print(r.get_area())
        print("--------------------")

    print("---------------------")
    shapes = [Circle(1), Rectangle(1, 1), Circle(20), Rectangle(2, 3)]
    for s in shapes:
        s.draw()
        print(s.get_area())
        print("--------------------")

if __name__ == "__main__":
    test()
```
