# counter 클래스 만들기
- 데이터: 현재 숫자
- 동작: 증가하기, 현재 값 읽어오기

```python
class Counter:
    def __init__(self, initial_count=0):
        self.count = initial_count

    def increment(self, amount=1):
        self.count += amount

    def get_count(self):
        return self.count


def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    for i in range(10):
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment(3)
    print(yourc.get_count())

if __name__ == "__main__":
    test_counter()
```

# TV 클래스 추가
- 데이터: 현재 채널번호, 볼륨량, ON 여부
- 동작: 켜기, 끄기, 채널변경, 볼륨변경, 현재 TV 객체 내용 출력

- 클래스명: MyTelevision

```python
class MyTelevision:
    def __init__(self, channel=1, volume=10, is_on=False):
        self.channel = channel
        self.volume = volume
        self.is_on = is_on

    def turn_on(self):
        self.is_on = True

    def turn_off(self):
        self.is_on = False

    def change_channel(self, channel):
        self.channel = channel

    def change_volume(self, volume):
        self.volume = volume

    def display_status(self):
        print(f"Channel: {self.channel}, Volume: {self.volume}, Is On: {'True' if self.is_on else 'False'}")

def test_television():
    mytv = MyTelevision(channel = 11) 
    mytv.turn_on()
    mytv.display_status()
    mytv.turn_off()
    mytv.display_status()


if __name__ == "__main__":
    test_television()
```

# TV 리모컨 클래스

- 데이터는 없고 동작만 있음
- 연관된 TV를 켜거나 꺼라는 동작만 지원
- TV 상태 표시 추가

```python
class TvRemocon:
    def __init__(self, tv:MyTelevision):
        self.tv = tv

    def press_on(self):
        self.tv.turn_on()

    def press_off(self):
        self.tv.turn_off()

    def press_display(self):
        self.tv.display_status()

def test_remocon():
    tv1 = MyTelevision()
    rc1 = TvRemocon(tv1)
    rc1.press_on()
    rc1.press_display()
    rc1.press_off()
    rc1.press_display()

if __name__ == "__main__":
    test_remocon()
```

# 정보은닉
- 가시성 처리 방법

- private: 변수나 함수이름을 __로 시작한다. 
    * 외부에서 접근이 불가능하다.

```python
class Student:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def display_info(self):
        print(f"Name: {self.__name}, Age: {self.__age}")

    Kim = Student("Kim", 21)
    Kim.display_info()
    print(Kim.__age)
```
