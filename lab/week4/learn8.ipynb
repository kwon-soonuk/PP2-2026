# counter 클래스 만들기
- 데이터: 현재 숫자
- 동작: 증가하기, 현재 값 읽어오기

```python
class Counter:
    def __init__(self, initial_count=0):
        self.count = initial_count

    def increment(self, amount=1):
        self.count += amount

    def get_count(self):
        return self.count


def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    myc.increment()
    print(myc.get_count())
    for i in range(10):
        myc.increment()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increment(3)
    print(yourc.get_count())

if __name__ == "__main__":
    test_counter()
```

- 우리 팀의 4명에 대한 객체를 생성하고 목록에 저장 : ( list 사용 )

```python
from datetime import datetime

class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def get_age(self):
        current_year = datetime.now().year
        return current_year - self.__age

    def __str__(self):
        return f"Name: {self.__name}, Age: {self.get_age()}"

def test_person():
    team_members = [Person("정혜승", 2007), Person("김준서", 2007), Person("조하늘", 2007), Person("권순욱", 2007)]

    for member in team_members:
        print(member)

if __name__ == "__main__":
    test_person()
```

# 대화 학습 정리와 추가 실습

## 1. 객체, 클래스, 인스턴스

객체는 서로 관련된 정보와 동작을 함께 다룬다.
- 속성: 현재 풍량, 켜짐 여부처럼 객체가 기억하는 정보
- 동작: 풍량 높이기, 전원 끄기처럼 객체가 수행하는 기능
- 클래스: 어떤 속성과 동작을 가질지 정의한 설계도
- 인스턴스: 그 클래스로 만든 개별 객체
- 변수: 객체를 가리키는 이름

`Fan`은 클래스이고 `Fan()`은 새 인스턴스를 만드는 호출이다.
`living = Fan()`에서 `living`은 새 객체를 가리킨다.
같은 클래스로 만들어도 각각의 인스턴스는 서로 다른 속성값을 가질 수 있다.

## 2. self, 생성자, 기본값과 키워드 인자

`self`는 인스턴스 메서드가 현재 다루는 객체를 받는 매개변수이다.
`room.increase(1)`에서는 `room`이 가리키는 객체가 `self`에 자동 전달된다.
`__init__`은 새 객체의 초기 상태를 설정하며 교재에서는 생성자라고 부른다.

`self.speed = speed`의 왼쪽은 객체의 속성, 오른쪽은 전달받은 매개변수이다.
인자를 생략하면 기본값을 사용하고, 전달하면 그 값을 사용한다.
키워드 인자는 `location="거실"`처럼 매개변수 이름을 지정한다.
여기서 `None`은 위치가 지정되지 않았음을 표현한다.

```python
class Fan:
    category = "선풍기"

    def __init__(self, speed=0, location=None):
        self.speed = speed
        self.location = location

    def increase(self, amount=1):
        self.speed += amount

    def getSpeed(self):
        return self.speed

    def setSpeed(self, speed):
        self.speed = speed

    def __str__(self):
        return f"풍량: {self.speed}"


living = Fan(location="거실")
room = Fan(3)
living.increase(1)
print("거실:", living.speed, living.location)
print("다른 선풍기:", room.speed, room.location)
```

## 3. 접근자, 설정자, return, print, __str__

- 접근자(getter)는 저장된 값을 읽어 반환한다.
- 설정자(setter)는 저장된 값을 바꾼다.
- 설정한 값이 자동으로 반환되는 것은 아니다. 반환은 `return`으로 정한다.
- `return`이 없는 함수의 호출 결과는 `None`이다.
- `print`는 화면에 출력하며 출력한 값을 자동으로 반환하지 않는다.
- `__str__`는 반드시 문자열을 반환한다. `str(obj)`나 `print(obj)`에서 사용된다.

`description = str(a)`에 저장되는 것은 그 시점에 만든 문자열이다.
이후 객체의 상태가 바뀌어도 저장해 둔 문자열은 자동으로 갱신되지 않는다.

```python
a = Fan(3)
description = str(a)
changed = a.setSpeed(5)

print("저장한 문자열:", description)
print("설정자 반환값:", changed)
print("현재 객체:", a)
```

## 4. 정보 은닉과 조건을 검사하는 설정자

밑줄 두 개를 붙인 속성은 외부에서 그 이름으로 바로 읽거나 쓰기 어렵다.
접근자와 설정자를 통해 사용하게 하면 값 검사 규칙을 모을 수 있다.
아래 예제는 풍량 0~3만 저장한다.
조건이 거짓이면 `if` 내부 대입문을 건너뛰므로 기존 값이 유지된다.
실제로 `pass`를 실행하는 것은 아니다.

```python
class LimitedFan:
    def __init__(self):
        self.__speed = 0

    def getSpeed(self):
        return self.__speed

    def setSpeed(self, speed):
        if 0 <= speed <= 3:
            self.__speed = speed


limited = LimitedFan()
for requested_speed in (2, 7, 0):
    limited.setSpeed(requested_speed)
    print(f"요청 {requested_speed} → 저장된 풍량 {limited.getSpeed()}")
```

## 5. 객체 참조, 참조 공유, is

`a = Fan(); b = Fan()`은 객체를 두 개 만든다.
`b = a`는 새 객체를 만들지 않고 같은 객체를 가리키게 한다.
공유하는 객체의 속성을 변경하면 다른 이름으로 읽어도 바뀐 상태가 보인다.

`is`는 같은 객체를 가리키는지 확인한다.
`==`는 값이 같은지를 비교하며, 동일한 객체인지 확인하는 것과는 다르다.

```python
shared_a = Fan()
shared_b = shared_a
separate_c = Fan()
shared_b.setSpeed(3)

print(shared_a.getSpeed(), shared_b.getSpeed(), separate_c.getSpeed())
print("a와 b가 같은 객체:", shared_a is shared_b)
print("a와 c가 같은 객체:", shared_a is separate_c)
```

## 6. 객체를 함수에 전달하기

함수의 매개변수도 전달받은 같은 객체를 가리킨다.
객체의 속성을 변경하는 것과, 매개변수에 새 객체를 대입하는 것은 다르다.
함수 안에서 `fan = Fan()`을 실행하면 그 이름만 새 객체를 가리킨다.
호출한 쪽 변수의 참조까지 바뀌지는 않는다.

```python
def change(fan):
    fan.setSpeed(1)
    fan = Fan()
    fan.setSpeed(3)


original = Fan()
alias = original
change(alias)
print(original.getSpeed(), alias.getSpeed(), original is alias)
```

## 8. 리스트를 속성으로 저장하기

생성자의 `self.topics = []`는 객체마다 새 리스트를 만든다.
`append`는 리스트에 항목을 추가하고 `len`은 항목 수를 구한다.

직접 작성하며 수정한 부분:
- `topics.append(topic)` → 현재 객체의 속성에 접근해야 한다.
- `self.len(topics)` → `len`은 내장 함수이고, 대상 리스트가 객체의 속성이다.
- 올바른 구조: `len(self.topics)`

```python
class StudyLog:
    def __init__(self):
        self.topics = []

    def addTopic(self, topic):
        self.topics.append(topic)

    def getCount(self):
        return len(self.topics)


my_log = StudyLog()
another_log = StudyLog()
my_log.addTopic("객체와 클래스")
my_log.addTopic("self와 생성자")
another_log.addTopic("반복문")
print("내 기록:", my_log.topics, "개수:", my_log.getCount())
print("다른 기록:", another_log.topics, "개수:", another_log.getCount())
```

## 9. 딕셔너리, 반복문, 설명 문자열

딕셔너리는 키와 값을 연결한다. 같은 키에 대입하면 기존 값을 갱신한다.
`self.scores[subject] = score`는 저장, `self.scores[subject]`는 조회이다.
조회할 키가 이미 저장되어 있다고 가정했다.

`.items()`는 키와 값을 한 쌍씩 꺼낸다.
문자열을 리스트에 모은 뒤 `"\n".join(lines)`으로 줄바꿈을 넣어 연결한다.
`__str__` 안에서는 직접 출력하지 않고 완성된 문자열을 반환한다.
f-string은 여는 따옴표와 닫는 따옴표가 모두 필요하다.

```python
class ScoreBook:
    def __init__(self):
        self.scores = {}

    def setScore(self, subject, score):
        self.scores[subject] = score

    def getScore(self, subject):
        return self.scores[subject]

    def __str__(self):
        lines = []
        for subject, score in self.scores.items():
            lines.append(f"{subject}: {score}")
        return "\n".join(lines)


book = ScoreBook()
book.setScore("수학", 80)
book.setScore("영어", 90)
print("수학 점수:", book.getScore("수학"))
print(book)
book.setScore("수학", 95)
print("수학 점수 변경 후:", book.getScore("수학"))
```
